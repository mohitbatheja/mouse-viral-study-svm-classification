```python
import pandas as pd
```

```python
df = pd.read_csv("mouse_viral_study.csv")
```

```python
df
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
x = df.drop("Virus Present",axis = 1)
y = df["Virus Present"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.20, random_state = 42
)
```

```python
scaler = StandardScaler()
```

```python
scaler
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
svm_c = SVC()

svm_c.fit(x_train_scaled,y_train)
```

```python
y_pred = svm_c.predict(x_test_scaled)
```

```python
y_pred
```

```python
compariosn = pd.DataFrame({
    "Actual" : y_test,
    "Prediction" : y_pred
})
```

```python
compariosn
```

```python
accuracy = accuracy_score(y_test,y_pred)*100
```

```python
print(f"accuracy:- {accuracy:.2f}%")
```

```python
classification = classification_report(y_test,y_pred)
```

```python
print("Classification Report:\n",classification)
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
print("confusion matrix:\n",cm)
```

**Conclusion**

The SVM classification model achieved 100% accuracy on the test dataset. All 80 test samples were correctly classified, with 42 samples from Class 0 and 38 samples from Class 1. Both classes achieved 100% precision, recall, and F1-score. The confusion matrix shows zero false positives and zero false negatives, indicating excellent classification performance on the test dataset.
